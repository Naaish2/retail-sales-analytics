# End-to-End Retail Sales Analytics Project

This notebook analyzes a Walmart-style retail sales dataset for revenue, profit, product, customer, regional, monthly trend, and discount impact insights.

## Business Questions

- Which products generate the most revenue?
- Which regions are most profitable?
- Which customer segments buy the most?
- Are discounts increasing or hurting profit?
- What are monthly sales trends?

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/processed/walmart_sales_clean.csv')
df['Date'] = pd.to_datetime(df['Date'])
df.head()
```

```python
kpis = {
    'total_sales': df['Revenue'].sum(),
    'total_profit': df['Profit'].sum(),
    'profit_margin': df['Profit'].sum() / df['Revenue'].sum(),
    'stores': df['Store'].nunique(),
    'products': df['Product'].nunique() if 'Product' in df.columns else np.nan,
}
kpis
```

```python
monthly_trend = df.groupby('Month', as_index=False).agg(
    total_revenue=('Revenue', 'sum'),
    total_profit=('Profit', 'sum')
)
monthly_trend.head()
```

```python
top_products = df.groupby(['Product', 'Category'], as_index=False).agg(
    total_revenue=('Revenue', 'sum'),
    total_profit=('Profit', 'sum'),
    average_discount=('Discount', 'mean'),
    average_margin=('Profit_Margin', 'mean')
).sort_values('total_revenue', ascending=False).head(10)
top_products
```

```python
region_performance = df.groupby('Region', as_index=False).agg(
    total_revenue=('Revenue', 'sum'),
    total_profit=('Profit', 'sum'),
    average_margin=('Profit_Margin', 'mean')
).sort_values('total_profit', ascending=False)
region_performance
```

```python
discount_impact = df.assign(
    discount_band=pd.cut(df['Discount'], bins=[-0.01, 0.03, 0.07, 0.12, 1.0], labels=['0-3%', '3-7%', '7-12%', '12%+'])
).groupby('discount_band', as_index=False, observed=False).agg(
    total_revenue=('Revenue', 'sum'),
    total_profit=('Profit', 'sum'),
    average_margin=('Profit_Margin', 'mean')
)
discount_impact
```

## Insights and Recommendations

- Compare high-revenue products against profit margin before increasing discounts.
- Use region and customer segment performance to prioritize marketing spend.
- Plan inventory around monthly and holiday demand peaks.
- Reduce promotions where discount bands show weak profit contribution.
